```python
import requests
import json

def format_resume(raw_text):
    prompt = f"""
    You are a professional resume formatter.
    Rewrite the following messy resume text into a clean recruiter-friendly format:
    {raw_text}
    """

    response = requests.post(
        "http://localhost:11434/api/generate",
        json={"model": "phi3", "prompt": prompt},
        stream=True,
    )

    formatted_text = ""
    raw_chunks = []  # store raw JSON chunks

    for line in response.iter_lines():
        if line:
            data = json.loads(line.decode("utf-8"))
            raw_chunks.append(data)  # keep raw JSON
            if "response" in data:
                formatted_text += data["response"]
            if data.get("done", False):
                break

    return formatted_text
```

```python
# Use this in a Jupyter notebook cell
from playwright.async_api import async_playwright

async def scrape_dynamic_job(url):
    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=False)
        page = await browser.new_page()
        await page.goto(url)
        print(await page.title())
        await browser.close()

# In Jupyter, you can directly await the function
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
```

```python
def get_driver(headless=True):
    options = Options()
    if headless:
        options.add_argument("--headless=new")
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")

    driver = webdriver.Chrome(
        service=Service(ChromeDriverManager().install()),
        options=options
    )
    return driver
```

```python
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

def scrape_job_description(url):
    driver = get_driver()
    driver.get(url)

    WebDriverWait(driver, 20).until(
        EC.presence_of_element_located((By.TAG_NAME, "body"))
    )

    job_text = driver.find_element(By.TAG_NAME, "body").text

    driver.quit()
    return " ".join(job_text.split())
```

```python
result = scrape_job_description("https://www.linkedin.com/jobs/search/?currentJobId=4346367604")
```
